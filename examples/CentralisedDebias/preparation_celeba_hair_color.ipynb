# Hair Color

```python
black_Hair = df["Black_Hair"]
blond_Hair = df["Blond_Hair"]
brown_Hair = df["Brown_Hair"]
gray_Hair = df["Gray_Hair"]
bald = df["Bald"]
```

```python
black_Hair = [0 if i == 1 else -1 for i in black_Hair]
blond_Hair = [1 if i == 1 else -1 for i in blond_Hair]
brown_Hair = [2 if i == 1 else -1 for i in brown_Hair]
gray_Hair = [3 if i == 1 else -1 for i in gray_Hair]
bald = [4 if i == 1 else -1 for i in bald]
```

```python
hair_color = [-2] * len(black_Hair)
multiple_colors = 0
for index, _ in enumerate(hair_color):
    num_colors = 0
    color = -2
    if black_Hair[index] != -1:
        color = black_Hair[index]
        num_colors += 1

    if blond_Hair[index] != -1:
        color = blond_Hair[index]
        num_colors += 1

    if brown_Hair[index] != -1:
        color = brown_Hair[index]
        num_colors += 1

    if gray_Hair[index] != -1:
        color = gray_Hair[index]
        num_colors += 1

    if bald[index] != -1:
        color = bald[index]
        num_colors += 1

    if num_colors > 1:
        color = -2
    hair_color[index] = color

    multiple_colors += 1 if num_colors > 1 else 0

print(f"There are {multiple_colors} people with multiple hair colors")
no_colors = Counter(hair_color)[-2]
print(f"There are {no_colors} people with no hair color")

counter_colors = Counter(hair_color)
print(f"People with black hair: {counter_colors[0]}")
print(f"People with blond hair: {counter_colors[1]}")
print(f"People with brown hair: {counter_colors[2]}")
print(f"People with gray hair: {counter_colors[3]}")
print(f"People BALD: {counter_colors[4]}")

print(
    f"Total people with hair color: {len(hair_color) - no_colors} - People with only a single hair color: {len(hair_color) - no_colors - multiple_colors} - Total Dataset size: {len(hair_color)}"
)
```
