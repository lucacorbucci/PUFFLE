```python
import numpy as np
```

```python
def update_lambda(current_lambda, D, T, prev_D=None, learning_rate=0.01, max_change=0.2):
    """
    Update lambda based on the difference between D and T, with improved scaling and trend consideration.

    :param current_lambda: Current value of lambda (float between 0 and 1)
    :param D: Current metric value
    :param T: Target metric value
    :param prev_D: Previous metric value (None if not available)
    :param learning_rate: Base rate at which lambda changes (default: 0.01)
    :param max_change: Maximum allowed change in lambda per update (default: 0.1)
    :return: Updated lambda value
    """
    # Calculate the relative difference (D - T) / T
    relative_diff = (D - T) / T

    # Apply a scaling function (e.g., tanh) to bound the difference
    scaled_diff = np.tanh(relative_diff)
    print(D, T, relative_diff, scaled_diff)

    # Determine direction of change
    if prev_D is not None:
        if D > T and D < prev_D:
            # D is decreasing but still above T, don't increase lambda
            direction = 0 if scaled_diff > 0 else -1
        elif D < T and D > prev_D:
            # D is increasing but still below T, don't decrease lambda
            direction = 0 if scaled_diff < 0 else 1
        else:
            direction = 1 if scaled_diff > 0 else -1
    else:
        direction = 1 if scaled_diff > 0 else -1

    # Calculate change in lambda
    delta_lambda = direction * learning_rate * abs(scaled_diff)

    # Limit the change to max_change
    delta_lambda = np.clip(delta_lambda, -max_change, max_change)

    # Update lambda
    new_lambda = current_lambda + delta_lambda

    # Ensure lambda stays within [0, 1]
    new_lambda = np.clip(new_lambda, 0, 1)

    return new_lambda
```

```python
# Example usage
lambda_value = 0.2  # Initial lambda value
target = 0.05  # Target value for D
prev_D = None

# Simulate the process over time
for t in range(100):
    # Simulate getting a new D value (replace this with your actual D calculation)
    D = np.random.normal(0.15, 2)  # Example: D fluctuates around 10 with some noise

    # Update lambda
    lambda_value = update_lambda(lambda_value, D, target, prev_D)

    print(f"Time {t}: D = {D:.2f}, lambda = {lambda_value:.4f}")

    prev_D = D  # Store current D for next iteration
```
