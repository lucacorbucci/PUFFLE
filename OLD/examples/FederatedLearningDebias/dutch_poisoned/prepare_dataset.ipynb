```python
from scipy.io import arff
import pandas as pd
import numpy as np
from collections import Counter
```

```python
data = arff.loadarff("../../../../data/dutch/dutch_census.arff")
```

```python
# convert to pandas dataframe
dutch_df = pd.DataFrame(data[0]).astype("int32")
```

```python
dutch_df.head()
```

```python
dutch_df.to_csv("../../../../data/dutch/dutch_census.csv", index=False)
```

```python
dutch_df["sex_binary"] = np.where(dutch_df["sex"] == 1, 1, 0)
dutch_df["occupation_binary"] = np.where(dutch_df["occupation"] >= 300, 1, 0)
del dutch_df["sex"]
del dutch_df["occupation"]
```

```python
dutch_df.head()
```

```python
Counter(dutch_df["edu_level"])
```

```python
dutch_df.shape
```

```python
gender_and_target = []
for gender, target in zip(list(dutch_df["edu_level"]), list(dutch_df["occupation_binary"])):
    gender_and_target.append((gender, target))

Counter(gender_and_target)
```

```python
gender_and_target = []
for gender, target in zip(list(dutch_df["sex_binary"]), list(dutch_df["occupation_binary"])):
    gender_and_target.append((gender, target))

Counter(gender_and_target)
```

```python
dutch_df.shape
```

```python
# drop rows from the dataframe where "sex_binary" is 0 and "occupation_binary" is 0
# Calculate the number of rows to drop
num_rows_to_drop = int(0.8 * len(dutch_df[(dutch_df["sex_binary"] == 0) & (dutch_df["occupation_binary"] == 0)]))

# Drop the rows
dutch_df = dutch_df.drop(
    dutch_df[(dutch_df["sex_binary"] == 0) & (dutch_df["occupation_binary"] == 0)].sample(num_rows_to_drop).index
)

# Verify the shape of the dataframe after dropping rows

dutch_df.shape
```

```python
gender_and_target = []
for gender, target in zip(list(dutch_df["sex_binary"]), list(dutch_df["occupation_binary"])):
    gender_and_target.append((gender, target))

Counter(gender_and_target)
```
