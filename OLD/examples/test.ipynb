```python
import numpy as np
```

```python
shap_1 = [0.5, 0.3, 0.8]
shap_2 = [0.2, 0.6, 0.4]
weights_f1 = [0.9, 0.7]
weights_faith = [0.1, 0]

avg = np.average([shap_1, shap_2], axis=0)
weighted_avg_f1 = np.average([shap_1, shap_2], weights=weights_f1, axis=0)
weighted_avg_faith = np.average([shap_1, shap_2], weights=weights_faith, axis=0)

print(avg)
print(weighted_avg_f1)
print(weighted_avg_faith)
```
